# Feature Interactions Under Distribution Shift

## 1. Introduction

Machine-learning models are commonly evaluated using test data that resembles their training data. However, measured model performance can change when the distribution of input features or the representation of different subgroups changes.

Understanding these changes is important because models may be evaluated on populations that differ from those represented in their training or testing data.

This project investigates how individual and combined changes in the evaluation weights of selected input-feature groups affect the predictive performance of machine-learning models on the UCI Adult Census Income dataset.

The project compares three classification algorithms: Logistic Regression, Decision Tree, and Random Forest. It evaluates their performance using accuracy and balanced accuracy under baseline and controlled reweighting conditions.

The experiments provide an initial investigation of subgroup reweighting. They do not fully simulate a naturally occurring distribution shift.

## 2. Research Question

How do individual and combined changes in input-feature distributions affect the predictive performance of machine-learning models?

In this initial study, the question is investigated through controlled reweighting of selected subgroups in the test dataset.

## 3. Project Objectives

The main objectives are:

1. Establish baseline performance using the original test dataset.
2. Examine the effects of reweighting selected feature groups individually and in combination.
3. Compare Logistic Regression, Decision Tree, and Random Forest.
4. Evaluate changes in accuracy and balanced accuracy.
5. Investigate how the strength of education-based reweighting affects measured performance.
6. Identify limitations and possible directions for further research.

## 4. Dataset and Experimental Methodology

### 4.1 Dataset

This project uses the UCI Adult Census Income dataset, which contains demographic and employment-related information used to predict whether an individual's annual income exceeds $50,000.

The dataset contains:

- Training records: 32,561
- Test records: 16,281
- Input features: 14
- Target variable: Income, classified as `<=50K` or `>50K`

The dataset contains both numerical and categorical features. Question-mark values representing missing information were identified in categorical columns, including workclass, occupation, and native country.

### 4.2 Data Preprocessing

The following preprocessing operations were used:

- Numerical features: missing values were imputed using the median, and features were standardized.
- Categorical features: missing values were imputed using the most frequent value, and categories were one-hot encoded.
- Preprocessing and classification were combined into a pipeline.

The models were trained using the original training dataset and evaluated on the original test records.

### 4.3 Baseline Models

Three classification algorithms were evaluated:

1. Logistic Regression
2. Decision Tree
3. Random Forest

The baseline measurements were calculated before applying the additional subgroup evaluation weights.

### 4.4 Controlled Reweighting Experiments

The experiments examined the effects of assigning greater evaluation weight to selected subgroups.

The feature conditions were:

- **Age:** age greater than or equal to 40.
- **Working hours:** more than 45 hours per week.
- **Education:** education number greater than 12.

The combined conditions were:

- Age + Working hours
- Age + Education
- Working hours + Education
- Age + Working hours + Education

In the main feature-group comparison, selected records received a weighting multiplier of 3. Records outside the selected subgroup retained their original relative weight before normalization.

Weights were normalized separately within each income class to preserve the original total weight assigned to each class. This ensured that the overall income-class proportions remained unchanged in the weighted evaluation.

A separate experiment examined education-based reweighting with multipliers of 2, 3, and 5, compared with the baseline multiplier of 1.

**Methodological distinction:** These experiments reweight existing test records during evaluation. The models were not retrained under new distributions, and new test populations were not generated. Consequently, the experiments investigate the sensitivity of weighted evaluation metrics to subgroup representation rather than fully reproducing real-world distribution shift.

### 4.5 Evaluation Metrics

Two primary metrics were used.

**Accuracy:** The proportion of predictions that are correct.

**Balanced accuracy:** The average recall across the two income classes. It gives equal importance to the recall of each class.

Both metrics were used because the dataset contains more observations in the `<=50K` class than in the `>50K` class. Examining both helps reveal changes that may not be apparent from accuracy alone.

## 5. Experimental Results

### 5.1 Baseline Performance

The baseline performance on the original test dataset was:

| Model | Accuracy | Balanced Accuracy |
|---|---:|---:|
| Logistic Regression | 85.07% | 75.97% |
| Decision Tree | 81.27% | 74.41% |
| Random Forest | 85.09% | 76.73% |

Random Forest achieved the highest baseline balanced accuracy at approximately 76.73%. Its accuracy was very close to that of Logistic Regression. Decision Tree recorded the lowest baseline accuracy and balanced accuracy among the three models.

These unweighted baseline measurements provide the reference values for comparison with the subsequent subgroup-reweighting experiments, in which sample weights are applied during evaluation.

### 5.2 Individual and Combined Feature Reweighting

The following table reports the change in balanced accuracy from each model's baseline under the multiplier-3 reweighting experiments.

| Feature condition | Logistic Regression | Decision Tree | Random Forest |
|---|---:|---:|---:|
| Age | +0.31 pp | −0.83 pp | +0.21 pp |
| Working hours | +0.83 pp | 0.00 pp | +0.38 pp |
| Education | +4.07 pp | +1.80 pp | +2.95 pp |
| Age + Working hours | +1.24 pp | +0.55 pp | +1.08 pp |
| Age + Education | +4.05 pp | +1.75 pp | +3.00 pp |
| Working hours + Education | +3.21 pp | +1.79 pp | +2.27 pp |
| Age + Working hours + Education | +2.82 pp | +1.67 pp | +2.13 pp |

*Note: pp means percentage points. Positive values indicate an increase in balanced accuracy relative to the corresponding baseline; negative values indicate a decrease. Values are rounded.*

Education-based reweighting produced the largest balanced-accuracy increase among the three individual feature conditions for all three models.

The size of the change differed across models. Logistic Regression showed the largest increase under education-based reweighting, at approximately 4.07 percentage points. Random Forest increased by approximately 2.95 percentage points, while Decision Tree increased by approximately 1.80 percentage points.

Age-based reweighting slightly reduced Decision Tree balanced accuracy, while producing small positive changes for the other two models.

### 5.3 Combined Feature Conditions

The combined-feature experiments did not consistently produce larger increases in balanced accuracy than education-only reweighting.

For example, Logistic Regression showed an increase of approximately 4.07 percentage points under education-only reweighting, compared with approximately 2.82 percentage points when all three feature conditions were combined.

Random Forest showed a similar pattern: the increase was approximately 2.95 percentage points for education alone and 2.13 percentage points for the combination of all three conditions.

These findings indicate that the effects of the selected reweighting conditions were not simply additive in these experiments. However, this comparison alone does not establish a formal statistical interaction between the features.

### 5.4 Education Reweighting Strength

A separate experiment investigated whether increasing the education reweighting multiplier was associated with changes in balanced accuracy.

| Model | Baseline (×1) | Multiplier 2 | Multiplier 3 | Multiplier 5 |
|---|---:|---:|---:|---:|
| Logistic Regression | 75.97% | 78.87% | 80.04% | 80.85% |
| Decision Tree | 74.41% | 75.74% | 76.21% | 76.44% |
| Random Forest | 76.73% | 78.86% | 79.69% | 80.24% |

Balanced accuracy increased at each tested multiplier for all three models.

At multiplier 5, the changes relative to baseline were approximately:

- Logistic Regression: +4.89 percentage points
- Decision Tree: +2.03 percentage points
- Random Forest: +3.51 percentage points

This pattern was observed for the tested education condition, multiplier values, models, and test dataset. It should not be assumed to hold for other datasets or reweighting schemes.

### 5.5 Weighted Accuracy Trade-Off at Multiplier 5

Although balanced accuracy increased, ordinary accuracy decreased at the strongest education reweighting level.

| Model | Baseline Accuracy | Accuracy at Multiplier 5 |
|---|---:|---:|
| Logistic Regression | 85.07% | 83.83% |
| Decision Tree | 81.27% | 79.78% |
| Random Forest | 85.09% | 83.90% |

These results show that weighted accuracy and weighted balanced accuracy can respond differently to changes in evaluation weights. Weighted balanced accuracy gives equal importance to recall for each class after accounting for the sample weights, while weighted accuracy measures the proportion of correct predictions under those weights.

Therefore, an increase in weighted balanced accuracy does not necessarily mean that every evaluation metric has improved. These findings describe the sensitivity of fixed model predictions to the chosen reweighting scheme, rather than a change in the models' learned predictive ability.

## 6. Discussion

The experiments demonstrate that the measured performance of a classifier can vary with the representation of selected subgroups in the evaluation procedure.

First, education-based reweighting produced the largest balanced-accuracy increase among the individual feature conditions tested for all three models. This suggests that the selected education subgroup had a stronger effect on the measured results than the selected age or working-hours subgroups under this particular weighting scheme.

Second, the effects differed across algorithms. Logistic Regression showed a larger balanced-accuracy increase under education-based reweighting than Decision Tree. Random Forest also showed a substantial increase. These differences indicate that the response to a given evaluation condition depends on the model.

Third, combining feature conditions did not consistently improve balanced accuracy beyond education-only reweighting. The results do not support the assumption that adding more selected feature groups will necessarily lead to a larger increase.

Fourth, stronger education reweighting increased balanced accuracy but reduced ordinary accuracy for all three models at multiplier 5. This highlights the importance of specifying the evaluation metric and the weighting procedure when comparing model performance.

These findings are preliminary. Because the experiments use a fixed test dataset with reweighted observations, they describe changes in weighted evaluation results rather than the full behavior of models under naturally occurring distribution shifts.

## 7. Limitations

The study has several limitations.

1. **Evaluation reweighting:** Existing test observations were reweighted rather than replaced with data from a genuinely shifted population.
2. **Single dataset split:** The experiments used one fixed training/test split. Results may differ across other splits or samples.
3. **Limited feature conditions:** Only selected age, working-hours, and education conditions were investigated.
4. **Limited models:** The comparison included Logistic Regression, Decision Tree, and Random Forest, but not other classification algorithms.
5. **Uneven metric coverage:** Accuracy and balanced accuracy were the primary metrics in the main model and feature-condition comparisons. Precision, recall, F1-score, and additional class-specific metrics were examined in a smaller set of experiments, so they do not provide equally comprehensive coverage across all conditions.
6. **Limited generalizability:** The experiments used one dataset, so the findings may not generalize to other datasets or populations.
7. **No causal conclusions:** The results do not establish that a feature itself causes changes in model performance.
8. **Historical dataset:** The Adult Census dataset represents historical census data. Its results should not automatically be generalized to current populations.
9. **No formal interaction test:** The combined-feature comparisons are descriptive and do not by themselves establish statistically significant interactions between features.

## 8. Conclusion

This project investigated how individual and combined feature-based evaluation reweighting affected the measured performance of Logistic Regression, Decision Tree, and Random Forest on the UCI Adult Census Income dataset.

Education-based reweighting produced the largest balanced-accuracy increase among the individual feature conditions tested for all three models. Combined feature conditions did not consistently outperform education-only reweighting, and the magnitude of the changes varied by model.

Increasing the education reweighting multiplier from 1 to 5 was associated with higher balanced accuracy for all three models. However, ordinary accuracy decreased at multiplier 5, demonstrating that conclusions about performance depend partly on the evaluation metric.

The main conclusion is that the measured performance of these fixed models depends on how observations are weighted during evaluation. These controlled reweighting experiments do not establish how the models would perform on a genuinely shifted population. Further research using independently sampled shifted populations, additional datasets, repeated experiments, and formal interaction analyses is needed to investigate the broader relationship between feature-distribution changes and model performance.

This study is an initial empirical investigation. It does not claim that the observed patterns are universal or that the project establishes a novel research contribution.

## 9. Future Work

The following directions could strengthen the investigation:

1. **Test genuine distribution shifts:** Evaluate models on test datasets whose feature distributions differ from those of the training data.
2. **Repeat the experiments:** Use multiple random seeds or data splits to investigate the stability of the findings.
3. **Investigate additional feature groups:** Examine other demographic and employment-related features using clearly defined conditions.
4. **Use additional evaluation metrics:** Compare precision, recall, F1-score, and class-specific performance alongside accuracy and balanced accuracy.
5. **Explore more weighting strengths:** Investigate how different weighting schemes affect the results.
6. **Test additional models:** Include other classification algorithms to assess whether the observed patterns depend on the chosen models.
7. **Investigate feature interactions formally:** Use suitable statistical or machine-learning methods to test whether combined feature effects differ from what would be expected from their individual effects.
8. **Evaluate additional datasets:** Determine whether the observations persist in other classification tasks and populations.
9. **Review related literature:** Compare the research question, methodology, and results with previous studies to identify what is already known and where a meaningful contribution might be possible.
